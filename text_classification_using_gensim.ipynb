```python
import pandas as pd
fakeNews=pd.read_csv("Fake.csv")
fakeNews.head()
```

```python
trueNews=pd.read_csv("True.csv")
trueNews.head()
```

```python
fakeNews['label']='fake'
fakeNews['label_num']=0
trueNews['label']='true'
trueNews['label_num']=1
```

```python
df=pd.concat([fakeNews,trueNews],axis=0)
df=df.sample(frac=1).reset_index(drop=True)
df.head()
```

```python
df.head()
```

```python
df.shape
```

```python
df.label.value_counts()
```

```python
import spacy
nlp=spacy.load("en_core_web_sm")

def preprocess_and_vectorizer(text):
  doc=nlp(text)

  filtered_tokens=[]
  for token in doc:
    if token.is_stop or token.is_punct:
      continue
    filtered_tokens.append(token.lemma_)
  return wv.get_mean_vector(filtered_tokens)
```

```python
preprocess_and_vectorizer("may be, im not the best person for this")
```

```python
#meke average vector for few words. its a numerical representation for few words
wv.get_mean_vector(["good","person"])
```

```python
df["vector"]=df["text"].apply(lambda text:preprocess_and_vectorizer(text))
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(
    df.vector.values,
    df.label_num,
    test_size=0.2,
    random_state=2022,
    stratify=df.label_num
)
```

```python
import numpy as np
X_train_2d=np.stack(X_train)
X_test_2d=np.stack(X_test)
X_train_2d[:4]
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import classification_report
clf=GradientBoostingClassifier()
clf.fit(X_train_2d,y_train)
y_pred=clf.predict(X_test_2d)
print(classification_report(y_test,y_pred))
```
